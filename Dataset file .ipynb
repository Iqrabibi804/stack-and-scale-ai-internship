```python
import urllib.request, zipfile, io, os, glob
import pandas as pd

url = 'https://archive.ics.uci.edu/static/public/320/student+performance.zip'
with urllib.request.urlopen(url) as r:
    zipfile.ZipFile(io.BytesIO(r.read())).extractall('.')

for z in glob.glob('**/*.zip', recursive=True):
    zipfile.ZipFile(z).extractall(os.path.dirname(z) or '.')

path = glob.glob('**/student-mat.csv', recursive=True)[0]
print('File Founded:', path)

df = pd.read_csv(path, sep=';')
print(df.shape)
```
